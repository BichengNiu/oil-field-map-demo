检查源文件SHA256、ID覆盖、分支几何、原始数值及单位、核查判断和精确的百万桶/日换算。无可靠密度时不从吨/年换成原油体积；系统值不按分段重复累加。

```python
from pathlib import Path
import os, sys, tempfile, urllib.request
ROOT = next(path for path in (Path.cwd(), Path.cwd().parent) if (path / "audit_oil_pipelines.py").exists())
sys.path.insert(0, str(ROOT))
from audit_oil_pipelines import audit
from build_pipeline_data import SOURCE_URL
local_source = os.environ.get("GOIT_RAW_SOURCE")
if local_source:
    result = audit(local_source)
else:
    with tempfile.TemporaryDirectory(prefix="pipeline-audit-") as temporary:
        raw = Path(temporary) / "goit.geojson"
        urllib.request.urlretrieve(SOURCE_URL, raw)
        result = audit(raw)
result
```
